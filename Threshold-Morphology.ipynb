```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
img = cv2.imread("burger.jfif")
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
plt.imshow(img_rgb)
plt.axis("off")
plt.show()
```

```python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
plt.imshow(gray, cmap="gray")
plt.title("Gray Scale Image")
plt.axis("off")
plt.show()
```

```python
_, thresh = cv2.threshold(
gray,
127,
255,
cv2.THRESH_BINARY
)
plt.imshow(thresh, cmap="gray")
plt.title("Global Threshold")
plt.axis("off")
plt.show()
```

```python
adaptive = cv2.adaptiveThreshold(
gray,
255,
cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
cv2.THRESH_BINARY,
11,
2
)
plt.imshow(adaptive, cmap="gray")
plt.title("Adaptive Threshold")
plt.axis("off")
plt.show()
```

```python
kernel = np.ones((5, 5), np.uint8)
print(kernel)
```

```python
erosion = cv2.erode(
thresh,
kernel,
iterations=1
)
plt.imshow(erosion, cmap="gray")
plt.title("Erosion")
plt.axis("off")
plt.show()
```

```python
dilation = cv2.dilate(
thresh,
kernel,
iterations=1
)
plt.imshow(dilation, cmap="gray")
plt.title("Dilation")
plt.axis("off")
plt.show()
```

```python
opening = cv2.morphologyEx(
thresh,
cv2.MORPH_OPEN,
kernel
)
plt.imshow(opening, cmap="gray")
plt.title("Opening")
plt.axis("off")
plt.show()
```

```python
closing = cv2.morphologyEx(
thresh,
cv2.MORPH_CLOSE,
kernel
)
plt.imshow(closing, cmap="gray")
plt.title("Closing")
plt.axis("off")
plt.show()
```
